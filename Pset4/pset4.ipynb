```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
alpha = np.linspace(0,10, 100)
rf = 10/3 * 1/ (1+ 3**(-alpha)+2**(-alpha))
plt.plot(alpha,rf)
plt.xlabel('alpha')
plt.ylabel('Risk Free Rate')
plt.title('Risk Free Rate as Alpha Increases')
plt.show()
```

as alpha increases our risk free rate increases but at a decreasing rate. Alpha is our risk aversion, thus we would prefer to smooth our consumption over time. Our elasticity of intertemporal consumption decreases, thus we prefer not to change our consumption as much. We need a higher risk free rate to change our consumption patterns at our given risk aversion.
