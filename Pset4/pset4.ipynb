## Problem Set 4 Code

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
```

### Problem 1(a)

```python
alpha = np.linspace(0,10, 100)
rf = 10/3 * 1/ (1+ 3**(-alpha)+2**(-alpha))
plt.plot(alpha,rf)
plt.xlabel('alpha')
plt.ylabel('Risk Free Rate')
plt.title('Risk Free Rate as Alpha Increases')
plt.show()
```

as alpha increases our risk free rate increases but at a decreasing rate. Alpha is our risk aversion, thus we would prefer to smooth our consumption over time. Our elasticity of intertemporal consumption decreases, thus we prefer not to change our consumption as much. We need a higher risk free rate to change our consumption patterns at our given risk aversion.

### Problem 1(b)

```python
p0 = 3/10 * (3**(1-alpha) +2**(1-alpha)+1)

eri = 1/3* (3/p0 +2/p0 +1/p0)

risk_prem = eri-rf
```

```python
plt.plot(alpha,risk_prem)
plt.xlabel('Alpha')
plt.ylabel('Risk Premium')
plt.title('Risk Premium as Function of Alpha')
plt.show()
```

As risk aversion increases our risk premium also increases. For the agent to hold the risky asset, the risk premium must increase the more risk averse they are. They need the expected return to be high enough to offset states where consumption will be lower.
